## 导入依赖

```python
import math
import numpy as np
from kyle_robot_toolbox.robot_arm.arm4dof_uservo import Arm4DoFUServo
```

## 创建机械臂

```python
# 创建机械臂
arm = Arm4DoFUServo(config_folder="./config", \
                    is_init_pose=True, is_debug=False)
```

## 末端控制(自动生成Pitch)

```python
p_tool = [180.0, 0, 50.0]
arm.set_tool_pose(p_tool)
```

```python
p_tool = [150, 50, 0]
arm.set_tool_pose(p_tool)
```

```python
p_tool = [180.0, 50.0, 20.0]
arm.set_tool_pose(p_tool)
```

## 获取当前末端位姿

```python
p_tool = [180, 0, 160]
arm.set_tool_pose(p_tool)
```

```python
# 返回的数据格式: [x6, y6, z6, pitch, roll]  
x, y, z, pitch = arm.get_tool_pose()
print(arm.get_joint_angle_list())
print(f"坐标: X {x:.1f}  Y {y:.1f} Z {z:.1f}")
print(f"俯仰角(Pitch) {np.degrees(pitch):.1f} deg")
```
