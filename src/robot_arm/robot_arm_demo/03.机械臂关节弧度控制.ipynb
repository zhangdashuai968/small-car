## 导入依赖

```python
import math
import numpy as np
from kyle_robot_toolbox.robot_arm.arm4dof_uservo import Arm4DoFUServo
```

## 机械臂创建

```python
# 创建机械臂
arm = Arm4DoFUServo(config_folder="./config", \
                    is_init_pose=True)
```

## 关节控制

```python
arm.set_joint_angle_list([0, -math.pi/2, 0, 0, 0], is_wait=True)
```

```python
arm.set_joint_angle_list([0, np.deg2rad(-135), np.deg2rad(60), np.deg2rad(120), 0], is_wait=True)
```

```python
arm.set_joint_angle_list([0, -3*math.pi/4, math.pi/3, math.pi/2, 0], is_wait=True)
```

```python
arm.set_joint_angle_list([0, -3*math.pi/4, math.pi/3, math.pi/2, math.pi/2], is_wait=True)
```

## 关节控制Minimum Jerk轨迹规划

```python
arm.set_joint_angle_list_soft([math.pi/2, -math.pi/2, math.pi/4, 0], T=1.0)
arm.set_joint_angle_list_soft([-math.pi/2, -math.pi/2, math.pi/3, math.pi/3], T=1.5)
arm.set_joint_angle_list_soft([0, -3*math.pi/4, math.pi/2, math.pi/3], T=1.0)
arm.set_joint_angle_list_soft([0, -math.pi/2, 0, 0], T=1.0)
arm.set_joint_angle_list_soft([0, -math.pi/3, 0, 0], T=1.0)
arm.set_joint_angle_list_soft([math.pi/2, -math.pi/2, 0, 0], T=1.0)
arm.home()
print("完成动作")
```
